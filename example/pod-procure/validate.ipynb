```python
def add_arv_hints():
    cwltool.command_line_tool.ACCEPTLIST_EN_RELAXED_RE = re.compile(r".*")
    cwltool.command_line_tool.ACCEPTLIST_RE = cwltool.command_line_tool.ACCEPTLIST_EN_RELAXED_RE
    supported_versions = ["v1.0", "v1.1", "v1.2"]
    
    with open("schema.yaml") as f:
        schema_content = f.read()
    
    for s in supported_versions:
        use_custom_schema(s, "https://dask.org/cwl", schema_content)
    
    cwltool.process.supportedProcessRequirements.extend([
        "https://dask.org/cwl#DaskGatewayRequirement"
    ])



parsed_args = argparse.Namespace(
    validate=True,
    workflow="example.cwl",
    custom_schema_callback=add_arv_hints,
)
```

```python
stream_out = StringIO()
stream_err = StringIO()

res = cwlmain(
    args=parsed_args,
    stdout=stream_out,
)
```
